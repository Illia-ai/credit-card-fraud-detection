```python
# Загрузка данных (готовый датасет)
url = "https://raw.githubusercontent.com/nsethi31/Kaggle-Data-Credit-Card-Fraud-Detection/master/creditcard.csv"

try:
    df = pd.read_csv(url)
    print("✅ Данные загружены успешно!")
    print(f"Размер данных: {df.shape}")
    print(f"Первые 5 строк:")
    display(df.head())
except Exception as e:
    print(f"❌ Ошибка загрузки: {e}")
    print("Попробуем альтернативный источник...")
    
    # Альтернативный источник (если первый не работает)
    url2 = "https://raw.githubusercontent.com/plotly/datasets/master/fraud.csv"
    df = pd.read_csv(url2)
    print("✅ Данные загружены из альтернативного источника!")
    print(f"Размер данных: {df.shape}")
    display(df.head())
```

```python
# Анализ целевой переменной (Class)
print("="*50)
print("АНАЛИЗ ДИСБАЛАНСА КЛАССОВ")
print("="*50)

# Количество нормальных и мошеннических транзакций
normal = df['Class'].value_counts()[0]
fraud = df['Class'].value_counts()[1]
total = len(df)

print(f"✅ Нормальные транзакции (Class=0): {normal:,}")
print(f"❌ Мошеннические транзакции (Class=1): {fraud:,}")
print(f"📊 Доля мошенничества: {fraud/total*100:.4f}%")
print(f"📊 Доля нормы: {normal/total*100:.4f}%")

# Визуализация дисбаланса
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# График 1: Круговая диаграмма
axes[0].pie([normal, fraud], labels=['Норма', 'Мошенничество'], 
            autopct='%1.4f%%', explode=[0, 0.1], 
            colors=['#2ecc71', '#e74c3c'], startangle=90)
axes[0].set_title('Распределение классов', fontsize=14)

# График 2: Столбчатая диаграмма
sns.countplot(x='Class', data=df, ax=axes[1], palette=['#2ecc71', '#e74c3c'])
axes[1].set_title('Количество транзакций по классам', fontsize=14)
axes[1].set_xlabel('Class (0=Норма, 1=Мошенничество)')
axes[1].set_ylabel('Количество')

plt.tight_layout()
plt.show()
```

```python
# Импортируем наш класс модели
import sys
sys.path.append('.')  # Добавляем текущую папку в путь

from src.pipeline import HybridAnomalyDetector
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, classification_report, confusion_matrix

print("✅ Класс модели импортирован!")
print("="*50)
print("ПОДГОТОВКА ДАННЫХ ДЛЯ ОБУЧЕНИЯ")
print("="*50)

# Подготовка данных (если ещё не сделали)
X = df.drop('Class', axis=1)
y = df['Class']

# Разделяем на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print(f"Размер обучающей выборки: {X_train.shape}")
print(f"Размер тестовой выборки: {X_test.shape}")
print(f"Доля мошенничества в обучении: {y_train.mean():.4%}")
print(f"Доля мошенничества в тесте: {y_test.mean():.4%}")

# Создаем и обучаем модель
print("\n" + "="*50)
print("ОБУЧЕНИЕ ГИБРИДНОЙ МОДЕЛИ")
print("="*50)

detector = HybridAnomalyDetector(contamination=0.05, encoding_dim=16)
detector.fit(X_train, y_train)

# Предсказание на тестовых данных
y_pred = detector.predict(X_test)

print("\n" + "="*50)
print("РЕЗУЛЬТАТЫ МОДЕЛИ")
print("="*50)

print(f"F1-Score: {f1_score(y_test, y_pred):.4f}")
print("\nМатрица ошибок:")
print(confusion_matrix(y_test, y_pred))

print("\nОтчет по классам:")
print(classification_report(y_test, y_pred, target_names=['Норма', 'Мошенничество']))
```
